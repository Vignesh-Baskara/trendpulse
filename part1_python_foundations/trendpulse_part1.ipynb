```python

# ============================================
# TASK 2: DATA LOADING FUNCTION
# ============================================

# Required columns for our analysis
required_columns = [
    "video_id",
    "trending_date",
    "title",
    "channel_title",
    "category_id",
    "publish_time",
    "views",
    "likes",
    "dislikes",
    "comment_count",
    "comments_disabled",
    "ratings_disabled",
    "video_error_or_removed"
]


def load_country(path, country, max_rows=None):

    # Read CSV file
    df = pd.read_csv(path, nrows=max_rows)

    # Verify required columns
    missing_columns = [
        col for col in required_columns
        if col not in df.columns
    ]

    if missing_columns:
        raise ValueError(
            f"{country} is missing columns: {missing_columns}"
        )

    # Add country information
    df["country"] = country

    # Convert DataFrame into list of dictionaries
    records = df.to_dict("records")

    # Return records
    return records


# Load all three countries
all_records = []

for country, path in COUNTRY_FILES.items():

    country_records = load_country(path, country)

    all_records.extend(country_records)

    print(f"{country}: {len(country_records)} records loaded")

print("\nTotal records loaded:", len(all_records))
print("Data type:", type(all_records))
print("First record:", all_records[0])
```

```python

# ============================================
# BUILD CATEGORY NAMES DICTIONARY
# ============================================

category_df = pd.read_csv(CATEGORY_FILE)

print("Category file columns:", category_df.columns.tolist())

# Identify category name column
name_column = next(
    (col for col in ["category_name", "category", "name", "title"]
     if col in category_df.columns),
    None
)

if "category_id" not in category_df.columns or name_column is None:
    raise ValueError("Check the category CSV column names.")

category_records = category_df.to_dict("records")

category_names = {}

for record in category_records:

    category_id = str(record["category_id"])

    category_names[category_id] = record[name_column]

print("\nCategory dictionary created successfully!")
print("Total categories:", len(category_names))

print("\nCategory names:")

for category_id, category_name in category_names.items():
    print(f"{category_id}: {category_name}")
```

```python

# ============================================
# TASK 3: DATA CLEANING
# ============================================

from datetime import datetime

# Track parsing failures
parse_failures = {
    "views": 0,
    "likes": 0,
    "dislikes": 0,
    "comment_count": 0,
    "comments_disabled": 0,
    "ratings_disabled": 0,
    "video_error_or_removed": 0,
    "trending_date": 0,
    "publish_time": 0
}


# Check missing values
def is_missing(value):
    return value is None or value != value or str(value).strip() == ""


# Convert numerical values into integers
def convert_integer(value, field):

    try:
        if is_missing(value):
            raise ValueError("Missing value")

        return int(float(value))

    except (ValueError, TypeError, OverflowError):
        parse_failures[field] += 1
        return 0


# Convert values into Boolean
def convert_boolean(value, field):

    try:
        if is_missing(value):
            raise ValueError("Missing value")

        if isinstance(value, bool):
            return value

        if str(value).strip().lower() in ["true", "1"]:
            return True

        if str(value).strip().lower() in ["false", "0"]:
            return False

        raise ValueError("Invalid Boolean")

    except (ValueError, TypeError):
        parse_failures[field] += 1
        return False


# Convert trending date
def convert_trending_date(value):

    try:
        if is_missing(value):
            raise ValueError("Missing date")

        # Dataset format: YY.DD.MM
        return datetime.strptime(str(value).strip(), "%y.%d.%m").date()

    except (ValueError, TypeError):
        parse_failures["trending_date"] += 1
        return None


# Convert publishing date
def convert_publish_time(value):

    try:
        if is_missing(value):
            raise ValueError("Missing date")

        # Handles ISO format and trailing Z
        date_string = str(value).strip().replace("Z", "+00:00")

        return datetime.fromisoformat(date_string)

    except (ValueError, TypeError):
        parse_failures["publish_time"] += 1
        return None
```

```python

# ============================================
# APPLY CLEANING TO ALL RECORDS
# ============================================

cleaned_records = []

numeric_fields = [
    "views",
    "likes",
    "dislikes",
    "comment_count"
]

boolean_fields = [
    "comments_disabled",
    "ratings_disabled",
    "video_error_or_removed"
]

for record in all_records:

    cleaned_record = record.copy()

    # Convert numerical fields
    for field in numeric_fields:
        cleaned_record[field] = convert_integer(
            record.get(field), field
        )

    # Convert Boolean fields
    for field in boolean_fields:
        cleaned_record[field] = convert_boolean(
            record.get(field), field
        )

    # Convert date fields
    cleaned_record["trending_date"] = convert_trending_date(
        record.get("trending_date")
    )

    cleaned_record["publish_time"] = convert_publish_time(
        record.get("publish_time")
    )

    cleaned_records.append(cleaned_record)


print("Data cleaning completed!")
print("Total original records:", len(all_records))
print("Total cleaned records:", len(cleaned_records))
```

```python

# ============================================
# SETS: UNIQUE VIDEO IDS BY COUNTRY
# ============================================

country_video_ids = {}

for country in COUNTRIES:

    country_video_ids[country] = {
        record["video_id"]
        for record in cleaned_records
        if record["country"] == country
    }

    print(
        country,
        "unique video IDs:",
        len(country_video_ids[country])
    )


# Videos that trended in multiple countries
shared_videos = set()

for country1 in COUNTRIES:

    for country2 in COUNTRIES:

        if country1 != country2:

            shared_videos |= (
                country_video_ids[country1]
                & country_video_ids[country2]
            )

print("\nVideos trending in multiple countries:", len(shared_videos))


# Videos unique to each country
for country in COUNTRIES:

    other_country_videos = set()

    for other_country in COUNTRIES:

        if other_country != country:
            other_country_videos |= country_video_ids[other_country]

    unique_videos = (
        country_video_ids[country] - other_country_videos
    )

    print(
        f"{country} videos unique to that country:",
        len(unique_videos)
    )
```

```python

# ============================================
# DICTIONARY: CHANNEL APPEARANCE COUNTER
# ============================================

channel_counter = {}

for record in cleaned_records:

    channel = record["channel_title"]

    if is_missing(channel):
        channel = "Unknown"

    if channel in channel_counter:
        channel_counter[channel] += 1
    else:
        channel_counter[channel] = 1


# Sort channels by appearances
sorted_channels = sorted(
    channel_counter.items(),
    key=lambda item: item[1],
    reverse=True
)

# Use slicing to get top 15
top_15_channels = sorted_channels[:15]

print("TOP 15 TRENDING CHANNELS")
print("-" * 55)

for channel, appearances in top_15_channels:

    print(f"{channel}: {appearances} appearances")
```

```python

# ============================================
# TASK 6: TITLE FEATURES
# ============================================

def title_features(title):

    # Handle missing titles
    if is_missing(title):
        title = ""

    words = str(title).split()

    uppercase_words = 0
    exclamation_marks = 0
    question_marks = 0

    # Iterate over characters using a for loop
    for character in title:

        if character == "!":
            exclamation_marks += 1

        elif character == "?":
            question_marks += 1

    # Use range() with index access
    for i in range(len(words)):

        word = words[i]

        if word.isupper() and any(char.isalpha() for char in word):
            uppercase_words += 1

    return {
        "uppercase_words": uppercase_words,
        "exclamation_marks": exclamation_marks,
        "question_marks": question_marks,
        "word_count": len(words)
    }


# Test the function
for record in cleaned_records[:5]:

    features = title_features(record["title"])

    print("\nTitle:", record["title"])
    print(features)
```

```python

# ============================================
# WHILE LOOP, BREAK AND CONTINUE
# ============================================

index = 0
first_viral_video = None

while index < len(cleaned_records):

    record = cleaned_records[index]

    index += 1

    # Skip records with missing titles
    if is_missing(record["title"]):
        continue

    # Stop when the first video above the threshold is found
    if record["views"] >= VIRAL_VIEWS_THRESHOLD:

        first_viral_video = record

        break


if first_viral_video is not None:

    print("First video above viral threshold:")
    print("Title:", first_viral_video["title"])
    print("Views:", first_viral_video["views"])
    print("Country:", first_viral_video["country"])

else:

    print("No video found above the viral threshold.")
```

```python

# ============================================
# TASK 7: RULE-BASED CLASSIFICATION
# ============================================

def classify_video(record):

    views = record.get("views", 0) or 0
    likes = record.get("likes", 0) or 0
    dislikes = record.get("dislikes", 0) or 0

    ratings_disabled = record.get("ratings_disabled", False)
    comments_disabled = record.get("comments_disabled", False)

    # Handle invalid or missing numerical values
    try:
        views = max(0, int(views))
        likes = max(0, int(likes))
        dislikes = max(0, int(dislikes))

    except (ValueError, TypeError, OverflowError):
        return "Unrated"

    # First handle ratings availability
    if ratings_disabled:
        return "Unrated"

    total_reactions = likes + dislikes

    # Avoid division by zero
    if total_reactions == 0:
        return "Unrated"

    if views == 0:
        return "Low Engagement"

    # Calculate like ratio
    like_ratio = likes / total_reactions

    # Nested classification conditions
    if views >= VIRAL_VIEWS_THRESHOLD:

        if like_ratio >= HIGH_LIKE_RATIO:
            return "Viral Hit"

        elif like_ratio <= LOW_LIKE_RATIO:
            return "Divisive"

        else:
            if comments_disabled:
                return "Limited Feedback"

            return "Steady Performer"

    elif views >= STEADY_VIEWS_THRESHOLD:

        if like_ratio <= LOW_LIKE_RATIO:
            return "Divisive"

        elif comments_disabled:
            return "Limited Feedback"

        else:
            return "Steady Performer"

    else:

        return "Low Engagement"


print("Classification function created successfully!")
```

```python

# ============================================
# TASK 8: CODE REFACTORING
# ============================================

def count_by_field(records, field, default="Unknown"):

    counts = {}

    for record in records:

        value = record.get(field)

        if is_missing(value):
            value = default

        if value in counts:
            counts[value] += 1

        else:
            counts[value] = 1

    return counts


# Reuse the function for channel appearances
channel_counts_refactored = count_by_field(
    cleaned_records,
    "channel_title"
)

# Reuse the same function for classification counts
classification_counts_refactored = count_by_field(
    classified_records,
    "classification"
)


print("TOP 15 CHANNELS AFTER REFACTORING")
print("-" * 45)

top_channels_refactored = sorted(
    channel_counts_refactored.items(),
    key=lambda item: item[1],
    reverse=True
)

for channel, count in top_channels_refactored[:15]:
    print(f"{channel}: {count}")


print("\nCLASSIFICATION COUNTS AFTER REFACTORING")

for label, count in classification_counts_refactored.items():
    print(f"{label}: {count}")
```


## Refactoring Explanation

Previously, similar dictionary-counting logic was written separately for channel appearances and classification labels.

I refactored this repeated logic into a reusable count_by_field() function. This reduces code duplication, improves readability and makes future modifications easier.
